```python
df = pd.read_csv("Saas_Churn_Raw_Dataset.csv")
```

```python
df = df.drop_duplicates()
```

```python
print(df.shape)
```

```python
df.columns = df.columns.str.title()
```

```python
df.head()
```

```python
df["Plan_Type"].unique()
```

```python
df["Country"].unique()
```

```python
df["Payment_Method"].unique()
```

```python
df["Payment_Method"] = df["Payment_Method"].str.replace("_", " ")
```

```python
cols_to_convert = ["Plan_Type", "Country", "Payment_Method"]
for col in cols_to_convert:
    df[col] = df[col].str.strip().str.title()
```

```python
df["Country"].unique()
```

```python
df["Country"] = df["Country"].replace({"Usa": "USA", "U.S.A": "USA", "Uk": "UK", "U.K": "UK", "Uae": "UAE"})
```

```python
df["Payment_Method"] = df["Payment_Method"].replace({"Upi": "UPI"})
```

```python
df["Plan_Type"].unique()
```

```python
df["Country"].unique()
```

```python
df["Payment_Method"].unique()
```

```python
print(df.isna().sum())
```

```python
df["Industry"] = df["Industry"].fillna("Unknown")
```

```python
df["Acquisition_Channel"] = df["Acquisition_Channel"].fillna("Unknown")
```

```python
df["Churn_Reason"].unique()
```

```python
churned_missing = (df["Customer_Status"] == "Churned") & (df["Churn_Reason"].isna())
df.loc[churned_missing, "Churn_Reason"] = "Not Recorded"
```

```python
df["Churn_Reason"] = df["Churn_Reason"].replace("Unknown", "Not Recorded")
```

```python
df["Churn_Reason"] = df["Churn_Reason"].fillna("Unknown")
```

```python
df["Churn_Reason"].value_counts()
```

```python
missing_hours = (df["Avg_Resolution_Hours"].isna()) & (df["Support_Tickets_90D"] > 0)
df.loc[missing_hours, "Avg_Resolution_Hours"] = 0
```

```python
print(df.isna().sum())
```

```python
df["Signup_Date"] = pd.to_datetime(
    df["Signup_Date"], format="mixed", errors="coerce"
)
```

```python
df["Churn_Date"] = pd.to_datetime(
    df["Churn_Date"], format="mixed", errors="coerce"
)
```

```python
print(df.dtypes)
```

```python
invalid_dates_df = df[df["Churn_Date"] < df["Signup_Date"]]
print(len(invalid_dates_df))
```

```python
df = df.drop(invalid_dates_df.index)
```

```python
print(df.shape)
```

```python
end_date = df["Churn_Date"].fillna(pd.Timestamp("2025-11-30"))
```

```python
df["Tenure_Months"] = (end_date.dt.year - df["Signup_Date"].dt.year) * 12 + (
    end_date.dt.month - df["Signup_Date"].dt.month
)
```

```python
df.loc[df["Tenure_Months"] < 0, "Tenure_Months"] = 0
```

```python
df["Signup_Date"] = df["Signup_Date"].dt.date
```

```python
df["Churn_Date"] = df["Churn_Date"].dt.date
```
